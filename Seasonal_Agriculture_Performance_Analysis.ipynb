### 9. Outlier Analysis

Outlier analysis is used to identify unusually high or low values in the dataset using statistical methods and visualizations. The identified values are examined to determine whether they are errors, valid observations or values that need further investigation, rather than removing them automatically.

### Univariate: Season Distribution (Pie Chart)

A pie chart is used to show the proportion of records belonging to each season. This helps understand the overall distribution of the dataset across different seasons.

```python
# Univariate: Season distribution (Pie Chart)

plt.figure(figsize=(7, 7))

df["Season"].value_counts().plot.pie(
    autopct="%1.1f%%",
    startangle=90,
    cmap="viridis"
)

plt.title("Distribution of Records by Season")
plt.ylabel("")  # Hide the default "Season" label
plt.tight_layout()
plt.show()
```

```python
# Outlier analysis using IQR method

outlier_summary = []

for col in numeric_columns:
    q1 = df[col].quantile(0.25)
    q3 = df[col].quantile(0.75)
    iqr = q3 - q1

    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr

    count = ((df[col] < lower) | (df[col] > upper)).sum()

    outlier_summary.append({
        "Column": col,
        "Lower_Bound": lower,
        "Upper_Bound": upper,
        "Outlier_Count": count,
        "Outlier_Percentage": round(count / len(df) * 100, 2)
    })

outlier_summary = pd.DataFrame(outlier_summary).sort_values(
    "Outlier_Count", ascending=False
)

outlier_summary
```

### 12. Seasonal Comparison

Seasonal comparison is used to compare important agricultural measures across different seasons using summary tables and suitable visualizations. This helps identify which seasons show better or lower performance in terms of yield, production, profit, water usage and other relevant factors.

```python
# General seasonal summary

season_metrics = df.groupby("Season").agg(
    Records=("Farm_ID", "count"),
    Average_Yield=("Yield_Tonnes_Ha", "mean"),
    Total_Production=("Production_Tonnes", "sum"),
    Average_Profit=("Profit_INR", "mean"),
    Total_Profit=("Profit_INR", "sum"),
    Average_Water_Used=("Water_Used_m3", "mean")
).round(2)

season_metrics
```
